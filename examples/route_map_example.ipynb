# Interactive route map -- Sevilla (Metro Sevilla + TUSSAM)

Loads the combined Sevilla GTFS feeds and builds the interactive
`pyGTFSHandler.maps.route_map` for a single service day.

```python
from pathlib import Path
import datetime

from pyGTFSHandler import Feed
from pyGTFSHandler.maps import route_map

DATA_DIR = Path("test_files/sevilla")
# Group stops within 100m of each other into a common parent_station
# (e.g. paired platforms on either side of a road) so the map's
# stop-click/timetable logic treats them as one station.
feed = Feed([DATA_DIR / "Metro_Sevilla", DATA_DIR / "TUSSAM"], stop_group_distance=100)
```

2026-07-06 is a plain Monday (not a weekend day, and not near any obvious
Spanish public holiday) that falls inside both feeds' `calendar.txt`
validity ranges, so it's a reasonable, defensible "typical weekday" pick
without needing an actual holiday-calendar lookup here.

```python
service_date = datetime.date(2026, 7, 6)
m = route_map(feed, service_date)
m.save("map.html")
```

## Conflicting `direction_id` inspector

A separate, self-contained map: every stop with at least one
`direction_id_issues=True` row (see `Shapes.assign_direction_ids` in
`pyGTFSHandler/models/shapes.py`) is drawn as a red marker; every other
stop is a small gray dot. Clicking a red stop opens a panel listing the
`route_id`s that conflict there. Clicking a route opens a second panel
showing that route's conflicting `shape_id`s at this stop, alongside the
*longest* (by `shape_total_distance`) `shape_id` of that route that does
*not* conflict at this stop -- drawn on the map in blue, with its full
stop sequence listed -- as a reference for what the route "should" look
like through here.

Built by `pyGTFSHandler.maps.conflict_map` -- see `pyGTFSHandler/maps/conflict_map.py`. Saved to `conflict_map.html` -- open that file directly in a browser (same as `map.html` above); it isn't rendered inline in this notebook.


```python
from pyGTFSHandler.maps import conflict_map

m = conflict_map(feed)
m.save("conflict_map.html")
```
